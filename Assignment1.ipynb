2)  Add New Job Postings (interactive)

```python
job_postings_add_count=int(input('How many job postings do you want to add? )'))

for i in range(job_postings_add_count):
   
    Role_to_add=input('Enter role :')
    Company_to_add=input('Enter Company :')
    Skills_to_add=input('EnterSkills :')

    next_value=job_data['Job_ID'][-1]+1
    job_data['Job_ID'].append(next_value)
    
    #job_data.update({'Role': },{'company':'Company_to_add'},{'skills':'Skills_to_add'})
    job_data['Role'].append(Role_to_add)
    job_data['Company'].append(Company_to_add)
    job_data['Skills'].append(Skills_to_add)
    
    
   
```

3)  Clean & Normalize Skills (text processing) 


```python
def clean_skills(skills_str):
    skill=str(skills_str)
    text=skill.lower()

    #text=skill.replace(',',' ')

    return text

job_data['Skills']=[clean_skills(skill) for skill in job_data['Skills']]
print(job_data['Skills'])
```

4)  Skill Extraction & Frequency 

```python
all_Skills=[]
    
for skill in job_data['Skills']:

    str_splitted= skill.split()
    all_Skills.extend(str_splitted)

skill_count={}

for skill in  all_Skills:
    if skill in skill_count:
        skill_count[skill]+=1
    else:
        skill_count[skill]=1



list=[]
for skill,count in skill_count.items():
    list.append([count,skill])


list.sort(reverse=True)


for a,b in tuple(list):
    print(f'{b}:{a}')
```

6) Skill Search Function (functions & conditional logic) 

```python
def find_jobs_by_skill(skill_query):

    
    query=skill_query.lower().strip()

    match=[]
    for i in range(len(job_data['Job_ID'])):

        current_skill=job_data['Skills'][i].lower().split()

        if query in current_skill:

           job_details=[job_data['Job_ID'][i],job_data['Company'][i],job_data['Role'][i]]
           match.append(job_details)
   

    if match:
        return match
    else:
        print('not found')



find_jobs_by_skill('python')
   

```
